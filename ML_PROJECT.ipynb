```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder



def preprocess_data(
      month_size=30,
      handle_unknown='ignore',
      include_colors=False,
      label_breeds=False,
      incude_breed=True
    ):

  print("Starting preprocessing...")

  train_df = pd.read_csv(TRAIN_FILE_PATH)
  test_df = pd.read_csv(TEST_FILE_PATH)
  irrelavant_cols_train = ['Name', 'Outcome Time']
  irrelavant_cols_shared = ['Id', 'Found Location', 'Date of Birth','Sex upon Intake', 'Sex_None', 'Can Sex_nan', 'Intake Time']

  if not include_colors:
    irrelavant_cols_shared.append('Color')
  else:
      one_hot_encoded = np.zeros((len(train_df), len(unique_colors)))

      for i, color in enumerate(train_df['Color']):
        found_colors = []
        if "/" in color:
          found_colors.extend([val.strip() for val in color.split("/")])
        else:
          found_colors.append(color.strip())

        for color_val in found_colors:
          if color_val in color_mapping:
            category = color_mapping[color_val]
            j = unique_colors.index(category)
            one_hot_encoded[i, j] = 1

      color_df = pd.DataFrame(one_hot_encoded, columns=unique_colors)
      train_df = pd.concat([train_df, color_df], axis=1)

      one_hot_encoded = np.zeros((len(test_df), len(unique_colors)))

      for i, color in enumerate(test_df['Color']):
        found_colors = []
        if "/" in color:
          found_colors.extend([val.strip() for val in color.split("/")])
        else:
          found_colors.append(color.strip())

        for color_val in found_colors:
          if color_val in color_mapping:
            category = color_mapping[color_val]
            j = unique_colors.index(category)
            one_hot_encoded[i, j] = 1

      color_df = pd.DataFrame(one_hot_encoded, columns=unique_colors)
      test_df = pd.concat([test_df, color_df], axis=1)


      irrelavant_cols_shared.append('Color')

  # Convert columns to datetime format
  train_df['Intake Time'] = pd.to_datetime(train_df['Intake Time'], format="%m/%d/%Y %I:%M:%S %p")
  test_df['Intake Time'] = pd.to_datetime(test_df['Intake Time'], format='mixed')

  train_df['Date of Birth'] = pd.to_datetime(train_df['Date of Birth'], format="%m/%d/%Y")
  test_df['Date of Birth'] = pd.to_datetime(test_df['Date of Birth'], format="%m/%d/%Y")


  # Calculate Age upon Intake
  train_df['Age upon Intake'] = (train_df['Intake Time'] - train_df['Date of Birth']).dt.days // month_size
  test_df['Age upon Intake'] = (test_df['Intake Time'] - test_df['Date of Birth']).dt.days // month_size

  # Split Sex upon Intake into two columns
  train_df[['Can Sex', 'Sex']] = train_df['Sex upon Intake'].str.split(' ', expand=True)
  test_df[['Can Sex', 'Sex']] = test_df['Sex upon Intake'].str.split(' ', expand=True)


  # Bucket Sprayed and Neutered into same category
  train_df['Can Sex'] = train_df['Can Sex'].replace('Spayed', 'Neutered')
  train_df['Can Sex'] = train_df['Can Sex'].replace('Unknown', np.nan)

  test_df['Can Sex'] = test_df['Can Sex'].replace('Spayed', 'Neutered')
  test_df['Can Sex'] = test_df['Can Sex'].replace('Unknown', np.nan)


  train_df['Sex'] = train_df['Sex'].replace(np.nan, None)
  test_df['Sex'] = test_df['Sex'].replace(np.nan, None)


  train_df['Intake Condition'] = train_df['Intake Condition'].apply(lambda x: intake_condition_mapping.get(x, None))
  test_df['Intake Condition'] = test_df['Intake Condition'].apply(lambda x: intake_condition_mapping.get(x, None))

  one_hot_list = ['Intake Type','Intake Condition', 'Animal Type','Can Sex', 'Sex']


  if incude_breed:
    if label_breeds:
      # Label encode Breeds and Outcome Type
      train_df['Breed'] = train_df['Breed'].astype('category').cat.codes
      train_df['Outcome Type'] = train_df['Outcome Type'].astype('category').cat.codes
      irrelavant_cols_shared.append('Outcome Type')
    else:
      one_hot_encoded = np.zeros((len(train_df), len(breed_categories)))

      for i, breed in enumerate(train_df['Breed']):
        found_breeds = []
        if "/" in breed:
          found_breeds.extend([val.strip() for val in breed.split("/")])
        elif "Mix" in breed:
          found_breeds.append(breed.replace("Mix", "").strip())
          found_breeds.append("Mix")
        else:
          found_breeds.append(breed.strip())

        for breed_val in found_breeds:
          if breed_val in breed_to_category:
            category = breed_to_category[breed_val]
            j = breed_categories.index(category)
            one_hot_encoded[i, j] = 1

      breed_df = pd.DataFrame(one_hot_encoded, columns=breed_categories)
      train_df = pd.concat([train_df, breed_df], axis=1)

      one_hot_encoded_test = np.zeros((len(test_df), len(breed_categories)))
      for i, breed in enumerate(test_df['Breed']):
        found_breeds = []
        if "/" in breed:
          found_breeds.extend([val.strip() for val in breed.split("/")])
        elif "Mix" in breed:
          found_breeds.append(breed.replace("Mix", "").strip())
          found_breeds.append("Mix")
        else:
          found_breeds.append(breed.strip())

        for breed_val in found_breeds:
          if breed_val in breed_to_category:
            category = breed_to_category[breed_val]
            j = breed_categories.index(category)
            one_hot_encoded_test[i, j] = 1

      breed_df_test = pd.DataFrame(one_hot_encoded_test, columns=breed_categories)
      test_df = pd.concat([test_df, breed_df_test], axis=1)

      irrelavant_cols_shared.append('Breed')
  else:
    irrelavant_cols_shared.append('Breed')

  # One hot encode
  encoder = OneHotEncoder(handle_unknown=handle_unknown, sparse_output=False)

  encoded_vals = encoder.fit_transform(train_df[one_hot_list])
  encoded_df = pd.DataFrame(encoded_vals, columns=encoder.get_feature_names_out(one_hot_list))

  train_df = pd.concat([train_df, encoded_df], axis=1)

  encoded_test = encoder.transform(test_df[one_hot_list])
  encoded_test_df = pd.DataFrame(encoded_test, columns=encoder.get_feature_names_out(one_hot_list))

  test_df = pd.concat([test_df, encoded_test_df], axis=1)

  irrelavant_cols_shared.extend(one_hot_list)

  # Extract features and class labels
  class_labels = train_df['Outcome Type']
  irrelavant_cols_train.append('Outcome Type')
    # Label columns
    # train_df['Outcome Type'] = train_df['Outcome Type'].astype('category').cat.codes

  feature_set = train_df.drop(columns= irrelavant_cols_shared + irrelavant_cols_train)
  test_feature_set = test_df.drop(columns= irrelavant_cols_shared)


  return feature_set, test_feature_set, class_labels

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MaxAbsScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

def get_scaled(pca_variation=0.8):
  x_train, x_test, y = preprocess_data(incude_breed = True, include_colors=True)

  x_train.head().to_csv("x_train.csv")
  x_test.head().to_csv("x_test.csv")
  print("Scaling data...")
  scaler = MaxAbsScaler()
  x_train_scaled = scaler.fit_transform(x_train)
  x_test_scaled = scaler.transform(x_test)

  # print("Performihn PCA on data...")
  # pca = PCA(n_components=pca_variation)
  # x_train_scaled = pca.fit_transform(x_train_scaled)
  # x_test_scaled = pca.transform(x_test_scaled)

  return x_train_scaled, x_test_scaled, y
```

```python
import pickle
from datetime import datetime
import pandas as pd

def make_achar(model_name, accuracy, pipeline):
    try:
      timestamp = datetime.now().strftime("%m-%d_%H-%M")
      file = f"{MODEL_FOLDER_PATH}/{model_name}_{accuracy:.4f}_{timestamp}.pkl"
      with open(file, "wb") as f:
          pickle.dump(pipeline, f)
      return "Model saved as: " + file
    except Exception as e:
      return f"Error saving model: {e} "


def generate_submission(class_ind_predictions, class_labels, model_type, accuracy, description):

    submission_df = pd.DataFrame({
        'Id': range(1, len(class_ind_predictions) + 1),
        'Outcome Type': [class_labels[idx] for idx in class_ind_predictions]
    })
    timestamp = datetime.now().strftime("%m-%d_%H-%M")



    submission_df.to_csv(f"{SUBMISSION_FOLDER_PATH}/{model_type}_{accuracy:.4f}%_{description}_{timestamp}.csv", index = False)

```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

def train_random_forest():
  # Get data
  X, x_test, y = get_scaled()

  # Hyperparameter grid
  param_grid = {
      'n_estimators': [100, 200, 400],
      'max_depth': [10, 20, 30],
      'min_samples_leaf': [1, 2, 4],
      'bootstrap': [True, False]
  }

  # Random Forest Model
  rf_model = RandomForestClassifier(n_estimators=400)

  # RandomizedSearchCV with 5-fold cross-validation
  print("Starting GridSearchCV...")
  grid_search = GridSearchCV(
      estimator=rf_model,
      param_grid=param_grid,
      cv=5,
      scoring='balanced_accuracy',
      n_jobs=3,
      verbose=1
  )

  # # 5-fold cross validation
  # print("starting 5-fold cross validation")
  # cv_scores = cross_val_score(grid_search, X, y, cv=5, scoring="balanced_accuracy", n_jobs=3, verbose=1)
  # mean_accuracy = cv_scores.mean()

  # # Results
  # print(f"Cross-validation accuracy scores: {cv_scores}")
  # print(f"Mean accuracy: {mean_accuracy:.4f}")

  grid_search.fit(X, y)

  # Best model and score
  best_rf_model = grid_search.best_estimator_
  best_score = grid_search.best_score_
  best_params = grid_search.best_params_

  print(f"Best cross-validation balanced accuracy: {best_score:.4f}")
  print(f"Best parameters: {best_params}")

  # Best parameters: {'bootstrap': True, 'max_depth': 20, 'min_samples_leaf': 2}


  best_rf_model.fit(X, y)

  result = make_achar("random_forest", best_score, best_rf_model)
  return result
```

## Neural Network(Tensorflow)


```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras import backend as K
from sklearn.model_selection import KFold
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import pandas as pd
from sklearn.metrics import balanced_accuracy_score, classification_report

# Enable GPU processing
gpu_devices = tf.config.experimental.list_physical_devices('GPU')
if gpu_devices:
    for device in gpu_devices:
        tf.config.experimental.set_memory_growth(device, True)
    print("GPU is available and enabled for training.")
else:
    print("No GPU detected, running on CPU.")


# Define the model function
def create_model(input_dims):
    model = keras.Sequential([
        layers.Input(shape=(input_dims, 1)),  # Input layer
        layers.Flatten(),  # Flatten to make it compatible with Dense layers
        # layers.Dense(512, activation='relu'),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.1),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.1),
        layers.Dense(64, activation='relu'),
        # layers.Dense(256, activation='relu'),
        layers.Dropout(0.1),
        layers.Dense(5, activation='softmax')  # Output layer with 5 classes
    ])
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=[
                      tf.keras.metrics.Recall(name='recall'),
                      tf.keras.metrics.Precision(name='precision'),
                      tf.keras.metrics.CategoricalAccuracy(name='accuracy')
                  ])
    return model


# 5-Fold Cross Validation with Class Weights
def train_neural_network(num_epochs=30, input_dims = 21):
    X, x_test, y = get_scaled()

    dummies = pd.get_dummies(y)

    y = dummies.values

    label_order = dummies.columns.tolist()

    kf = KFold(n_splits=5, shuffle=True)
    cv_balanced_accuracies = []

    # Compute class weights
    class_weights = compute_class_weight('balanced', classes=np.unique(y.argmax(axis=1)), y=y.argmax(axis=1))
    class_weight_dict = dict(enumerate(class_weights))

    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        print(f"\nTraining Fold {fold + 1}...")

        X_train, X_val = X[train_index], X[val_index]
        y_train, y_val = y[train_index], y[val_index]

        with tf.device('/GPU:0'):
            model = create_model(input_dims)
            model.fit(
                X_train, y_train,
                epochs=num_epochs, batch_size=256, verbose=0,
                class_weight=class_weight_dict
            )

            # Predict class labels for validation set
            y_pred = model.predict(X_val)
            y_pred_classes = np.argmax(y_pred, axis=1)  # Convert probabilities to class labels
            y_true_classes = np.argmax(y_val, axis=1)

            # Compute balanced accuracy
            balanced_acc = balanced_accuracy_score(y_true_classes, y_pred_classes)
            cv_balanced_accuracies.append(balanced_acc)

            print(f"Fold {fold + 1} Balanced Accuracy: {balanced_acc:.4f}")

            report = classification_report(
                y_true_classes, y_pred_classes, target_names=label_order, output_dict=True
            )

            print(report)


    # Compute Mean Balanced Accuracy
    mean_balanced_accuracy = np.mean(cv_balanced_accuracies)
    print(f"\nCross-validation balanced accuracy scores: {cv_balanced_accuracies}")
    print(f"Mean Balanced Accuracy: {mean_balanced_accuracy:.4f}")

    print("\nTraining Final Model on Full Dataset...")
    with tf.device('/GPU:0'):
      final_model = create_model(input_dims)
      final_model.fit( X, y, epochs=num_epochs, batch_size=256, verbose=1, class_weight=class_weight_dict)

      predictions = final_model.predict(x_test)

    class_indices = np.argmax(predictions, axis=1)

    generate_submission(class_indices, label_order, "tensorflow", round(mean_balanced_accuracy*100,2), f"{num_epochs}_epochs_256_nodes_color_intake_condition_grouped")


    return  mean_balanced_accuracy, final_model

```

## XGBOOST

```python
import xgboost as xgb
from sklearn.model_selection import KFold
from sklearn.metrics import classification_report, balanced_accuracy_score
import pandas as pd
import numpy as np

def train_xgboost(num_folds=5):
    X, x_test, y = get_scaled()

    dummies = pd.get_dummies(y)
    y_encoded = dummies.values.argmax(axis=1)  # Convert one-hot to class index
    label_order = dummies.columns.tolist()

    kf = KFold(n_splits=num_folds, shuffle=True)
    fold_reports = []
    balanced_accuracies = []

    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        print(f"\nTraining Fold {fold + 1}...")

        X_train, X_val = X[train_idx], X[val_idx]
        y_train, y_val = y_encoded[train_idx], y_encoded[val_idx]

        model = xgb.XGBClassifier(
            objective='multi:softprob',
            num_class=len(label_order),
            eval_metric=balanced_accuracy_score,
            n_jobs=-1
        )

        model.fit(X_train, y_train)
        y_pred = model.predict(X_val)

        # Compute balanced accuracy
        balanced_acc = balanced_accuracy_score(y_val, y_pred)
        balanced_accuracies.append(balanced_acc)
        print(f"Fold {fold + 1} Balanced Accuracy: {balanced_acc:.4f}")

        # Classification report for recall
        report = classification_report(
            y_val, y_pred, target_names=label_order, output_dict=True
        )

        print("Recall per class:")
        for label in label_order:
            recall = report[label]["recall"]
            print(f"{label}: {recall:.4f}")

        fold_reports.append(report)

    print("\nMean Balanced Accuracy Across Folds: {:.4f}".format(np.mean(balanced_accuracies)))
    return fold_reports, balanced_accuracies
```

## AdaBoost Classifier

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

def train_adaboost_classifier(num_estimators):
    # Get data
    X, x_test, y = get_scaled()

    # Base estimator: shallow decision tree (stump)
    base_estimator = DecisionTreeClassifier(max_depth=1)

    # AdaBoost model
    ada_clf = AdaBoostClassifier(
        estimator=base_estimator,
        n_estimators=num_estimators,
        learning_rate=0.7,
    )

    print("Training AdaBoost Classifier with 5-fold CV...")
    cv_scores = cross_val_score(ada_clf, X, y, cv=5, scoring="balanced_accuracy", verbose=2)
    mean_accuracy = np.mean(cv_scores)

    print(f"Cross-validation accuracy scores: {cv_scores}")
    print(f"Mean accuracy: {mean_accuracy:.4f}")

    # Fit full model
    ada_clf.fit(X, y)

    # Save and return result
    result = make_achar("adaboost_classifier", mean_accuracy, ada_clf)
    return result
```

```python
import pickle
import os
import numpy as np
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.utils import Bunch
from sklearn.base import BaseEstimator, ClassifierMixin


# Extract accuracy "models/{name}_{accuracy:.4f}_{date}_{time}.pkl"
def extract_accuracy_from_name(filename):
    try:
        return float(filename.split("/_")[-3])
    except Exception:
        return 1.0  # Fallback/default if parsing fails


def load_model(path):
    with open(path, 'rb') as f:
        model = pickle.load(f)

    if hasattr(model, 'predict') and not hasattr(model, 'get_params'):
        return KerasWrapper(model)
    return model

def train_voting_ensemble(model_paths, name, weights=[]):
    from sklearn.preprocessing import LabelEncoder

    X, x_test, y = get_scaled()
    le = LabelEncoder()
    y = le.fit_transform(y)

    # Load all models
    models = []

    for i, path in enumerate(model_paths):
        model = load_model(f'{DATA_FOLDER_PATH}/{path}')
        models.append((f"model_{i}", model))

        # Weight based on accuracy extracted from filename
        if len(weights) != len(model_paths):
          acc = extract_accuracy_from_name(os.path.basename(path))
          weights.append(acc)

    # VotingClassifier
    voting_clf = VotingClassifier(
        estimators=models,
        voting='hard',
        # weights=weights
    )

    print("Training Voting Ensemble with 5-fold CV...")
    cv_scores = cross_val_score(voting_clf, X, y, cv=5, scoring="balanced_accuracy", n_jobs=3, verbose=2)
    mean_accuracy = np.mean(cv_scores)

    print(f"Cross-validation accuracy scores: {cv_scores}")
    print(f"Mean voting accuracy: {mean_accuracy:.4f}")

    # Fit full model
    voting_clf.fit(X, y)

    result = make_achar("voting_ensemble", mean_accuracy, voting_clf)
    return result
```
